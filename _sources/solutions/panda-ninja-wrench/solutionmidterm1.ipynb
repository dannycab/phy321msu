### Part 1, Particle in a one-dimensional  potential

We consider a particle (for example an atom) of mass $m$ moving in a one-dimensional potential,

$$
V(x)=\frac{V_0}{d^4}\left(x^4-2x^2d^2+d^4\right).
$$

We will assume all other forces on the particle are small in comparison, and neglect them in our model.  The parameters $V_0$ and $d$ are known constants. 

1. (5pt) Sketch or plot the potential and find the equilibrium points (stable and unstable) by requiring that the first derivative of the potential is zero. Make an energy diagram (see for example Malthe-Sørenssen chapter 11.3) and mark the equilibrium points on the diagram and characterize their stability. The position of the particle is $x$. 

```{admonition} Solution
:class: hint

To find the equilibrium points, we need to set the first derivative of the potential equal to zero and solve for $x$. This gives us the following equation:

$$\dfrac{dV(x)}{dx} = 4V_0\left(\dfrac{x^3}{d^4}-\dfrac{x}{d^2}\right) = 0.$$

$$ \dfrac{x}{d^2}\left(\dfrac{x^2}{d^2}-1\right) = 0.$$

We can solve this equation for $x$ and we find the equilibrium points.

$$x = 0, \quad x = \pm d.$$

To characterize their stability, we need to look at the second derivative of the potential. If the second derivative is positive, the equilibrium point is stable, if it is negative, the equilibrium point is unstable. The second derivative of the potential is given by:

$$\dfrac{d^2V(x)}{dx^2} = 4V_0\left(\dfrac{3x^2}{d^4}-\dfrac{1}{d^2}\right).$$

Evaluating those at each of the equilibrium points gives,

$$\dfrac{d^2V(0)}{dx^2} < 0, \quad \dfrac{d^2V(d)}{dx^2} > 0, \quad \dfrac{d^2V(-d)}{dx^2} > 0.$$

So that $x=0$ is an unstable equilibrium point and $x = \pm d$ are stable equilibrium points.

The code below plots $V(x)$ and locates the equilibrium points.

```

````python
def Potential(x):
    return V0/(d**4)*(x**4 - 2*x**2*d**2 + d**4)

xlim = 1.5
steps = 1000
V0 = 1.0
d = 1.0

x = np.linspace(-xlim, xlim, steps)
V = Potential(x)

fig = plt.figure(figsize=(8, 6))
plt.plot(x, V,lw=2, label=r'$V(x)$')

plt.plot(0,Potential(0),'C1o', label='Unstable')
plt.plot(-d,Potential(-d),'C2o', label='Stable')
plt.plot(d,Potential(d),'C2o')

plt.xlabel('x (m)')
plt.ylabel('V (J)')
plt.legend()
plt.grid(True)
````

```{admonition} Solution
:class: tip

This system conserves energy, so that at any point the total energy is constant. The total energy of the particle is given by $E=V(x)+K(v)$ at any point $x$. We can use this to find the velocity of the particle at $x=d$.

The total energy of the particle at $x=2d$ is given by,

$$E_{tot} = V(d) + K(v) = \dfrac{V_0}{d^4}\left(d^4-2d^2d^2+d^4\right) + \dfrac{1}{2}mv^2$$
$$E_{tot} = 9V_0= \dfrac{1}{2}mv^2$$

So that,

$$v = \sqrt{\dfrac{18V_0}{m}}.$$
```

5. (5pt) Use the above potential to find the net force acting on the particle.  Find the acceleration acting on the particle. Is this net force conservative? Calculate the **curl** of the force  $\boldsymbol{\nabla}\times \boldsymbol{F}$ in order to validate your conclusion. 

```{admonition} Solution
:class: tip

With a potential given by $V(x) = \frac{V_0}{d^4}\left(x^4-2x^2d^2+d^4\right)$, the net force acting on the particle is given by the negative gradient of the potential. We know this is one dimensional, so that the force is given as it was in part 1,

$$F(x) = -\dfrac{dV(x)}{dx} = -4V_0\left(\dfrac{x^3}{d^4}-\dfrac{x}{d^2}\right).$$

The acceleration acting on the particle is given by $a = F/m$,

$$a(x) = -\dfrac{4V_0}{m}\left(\dfrac{x^3}{d^4}-\dfrac{x}{d^2}\right).$$

The force is conservative because it is derived from a potential. We can also check this by calculating the curl of the force. The curl of the force is given by,

$$\boldsymbol{\nabla}\times \boldsymbol{F}$$

Because this is a one-dimensional problem, the only relevant components of the curl are those in $y$ and $z$, but they result from derivatives in $y$ and $z$, which are zero.

$$\boldsymbol{\nabla}\times \boldsymbol{F} = +\dfrac{dF(x)}{dz}\hat{y} - \dfrac{dF(x)}{dy}\hat{z} = 0.$$
```

6. (5pt) Are linear momentum and angular momentum conserved? You need to show this by calculating these quantities.

```{admonition} Solution
:class: tip

The linear momentum conservation is related to the net force acting on the particle. In three dimensions, this is given by:

$$\dfrac{d\boldsymbol{p}}{dt} = \boldsymbol{F}.$$

Here, momentum is conserved in $y$ and $z$ simply because the force is zero in those directions. In the $x$ direction, the force is given by $F(x) = -\dfrac{dV(x)}{dx}$, so that we have a force acting on the particle. This means that the linear momentum is not conserved.

With respect to the angular momentum, we can calculate it as $\boldsymbol{L} = \boldsymbol{r}\times\boldsymbol{p}$. The angular momentum is given by,

$$\boldsymbol{L} = \boldsymbol{r}\times\boldsymbol{p} = x\hat{x} \times p\hat{x} = 0.$$

Because we have a one-dimensional problem, the angular momentum is zero. This also means that the angular momentum is conserved in this case as well, because it is zero at all times. Why? There's no torque acting on the particle, so that the angular momentum is conserved.

$$\dfrac{d\boldsymbol{L}}{dt} = \boldsymbol{r}\times\boldsymbol{F} = x\hat{x} \times F(x)\hat{x} = 0.$$
```



7. (10pt) Write a numerical algorithm to find the position and velocity of the particle at a time $t+\Delta t$ given the position and velocity at a time $t$. Here you can use either the standard forward Euler, or the Euler-Cromer or the Velocity Verlet algorithms.   You need to justify your choice here (hint: consider energy conservation).

```{admonition} Solution
:class: tip

To start we need to write the equations of motion for the particle. The equations of motion are given by,

$$F(x) = m\ddot{x} = -\dfrac{dV(x)}{dx} = -4V_0\left(\dfrac{x^3}{d^4}-\dfrac{x}{d^2}\right)$$

so that,

$$\ddot{x} = -\dfrac{4V_0}{m}\left(\dfrac{x^3}{d^4}-\dfrac{x}{d^2}\right).$$

We need to seperate this into two first order differential equations. We can do this by defining the velocity of the particle as $\dot{x} = v$, so that,

$$\dot{x} = v,$$
$$\dot{v} = -\dfrac{4V_0}{m}\left(\dfrac{x^3}{d^4}-\dfrac{x}{d^2}\right).$$

This leads to the following Euler-Cromer step,

$$v_{i+1} = v_i - \dfrac{4V_0}{m}\left(\dfrac{x_i^3}{d^4}-\dfrac{x_i}{d^2}\right)\Delta t,$$
$$x_{i+1} = x_i + v_{i+1}\Delta t.$$

Below we implement this algorithm in Python. We use Euler-Cromer because we expect oscillatory motion, and we want to conserve energy. 

```


````python
def integrateEOM(V0, m, d, x0, v0, t0, tf, dt):
    t = np.arange(t0, tf, dt)
    x = np.zeros(t.shape)
    v = np.zeros(t.shape)
    x[0], v[0] = x0, v0

    for i in range(1, len(t)):
        v[i] = v[i-1] - dt * 4*V0/m*(x[i-1]**3/d**4 - x[i-1]/d**2)
        x[i] = x[i-1] + dt * v[i]

    return t, x, v
````

```{admonition} Solution
:class: tip

If we remove the $x^4$ term from the potential, we can see that the motion is different. The potential is given by,

$$V(x) = \dfrac{V_0}{d^4}\left(-2x^2d^2+d^4\right).$$

So that the force acting on the particle is given by,

$$F(x) = -\dfrac{dV(x)}{dx} = 4V_0\left(\dfrac{x}{d^2}\right).$$

We have a new set of equations of motion,

$$\ddot{x} = +\dfrac{4V_0}{m}\left(\dfrac{x}{d^2}\right),$$

And the Euler-Cromer step is given by,

$$v_{i+1} = v_i + \dfrac{4V_0}{m}\left(\dfrac{x_i}{d^2}\right)\Delta t,$$
$$x_{i+1} = x_i + v_{i+1}\Delta t.$$

We can program that and model the motion. But we should notice we've lost the confinement part of the potential. We expect the particle to run off to infinity. We can plot figures showing shorter and shorter times. This illustrates the blow up of the motion. You can see the motion is unbounded and the particle runs off to infinity.  
```

````python
def integrateEOMnoConfine(V0, m, d, x0, v0, t0, tf, dt):
    t = np.arange(t0, tf, dt)
    x = np.zeros(t.shape)
    v = np.zeros(t.shape)
    x[0], v[0] = x0, v0

    for i in range(1, len(t)):
        v[i] = v[i-1] + dt * 4*V0/m*(x[i-1]/d**2)
        x[i] = x[i-1] + dt * v[i]

    return t, x, v

m = 1
V0 = 1
d = 0.1

t0 = 0
tf = 30
N = 10000
dt = (tf - t0)/N

end = int(np.floor(N/100))

short=True
vel=False

initial_conditions = pd.DataFrame({'x0': [d,d,d], 'v0': [0.5, 1.5, 2.5]})

fig = plt.figure(figsize=(8, 6))
plt.xlabel('t (s)')

if vel:
    plt.title('Velocity of the particle')
    plt.ylabel('v (m/s)')
else:
    plt.title('Position of the particle')
    plt.ylabel('x (m)')

for i in range(len(initial_conditions)):
    x0 = initial_conditions.x0[i]
    v0 = initial_conditions.v0[i]
    t, x, v = integrateEOMnoConfine(V0, m, d, x0, v0, t0, tf, dt)

    if short:
        plt.plot(t[0:end], x[0:end], label='Trial {} (short time)'.format(i+1))
    else:
        plt.plot(t, x, label='Trial {}'.format(i+1))
        
plt.legend()

````

### EXAMPLE SOLUTION: A model of quark confinement

```{admonition} Solution
:class: tip

In part 2, we are going to study a new potential. This potential contains the basic mathematical components needed to describe the confinement of quarks through a term $\kappa x$. The aim of this exercise is to try to develop your intuition about the motion of physical objects due to specific forces. Then we will test our intuition by running simulations. The potential, in one dimension only is defined for $x\in [0.2,\infty)$.  It reads:

$$
V(x) = -\frac{\gamma}{x}+\frac{\delta}{x^2}+\kappa x,
$$

where the last term is the one which ensures confinement of for example quarks. We can show that this potential leads to an energy conserving force by calculating the **curl** of the resulting force.

This potential is $V(x) = -\frac{\gamma}{x}+\frac{\delta}{x^2}+\kappa x$. The force is given by the negative gradient of the potential, so that,

$$F(x) = -\dfrac{dV(x)}{dx} = -\dfrac{\gamma}{x^2}+\dfrac{2\delta}{x^3}-\kappa.$$

This is again a 1D force, so that the curl will be zero. We can show this explicitly by calculating the curl,

$$\boldsymbol{\nabla}\times \boldsymbol{F} = +\dfrac{dF(x)}{dz}\hat{y} - \dfrac{dF(x)}{dy}\hat{z} = 0.$$

Thus we expect the force to conserve energy, and because it is the only force acting on the particle, we expect the total energy to be conserved.

We also notice that the force is non-linear, so that the potential meets the criteria. 
```

2. (5pt) Plot the potential for $x\in [0.2,10]$ and set $\gamma=10$, $\delta = 3$ and $\kappa =1$. Find the value of $x$ where the potential has a minimum.

```{admonition} Solution
:class: tip

We can plot the potential using the given parameters. It might be hard to compute the location of the energy minimum sometimes. Graphing is a helpful tool to see if we are on the right track.

We can see from the plot that there is a minimum in the potential. We can find the location of the minimum by setting the first derivative of the potential equal to zero and solving for $x$. This gives us the following equation,

$$\dfrac{dV(x)}{dx} = 0 = -\dfrac{\gamma}{x^2}+\dfrac{2\delta}{x^3}-\kappa.$$

$$\dfrac{\gamma}{x^2} = \dfrac{2\delta}{x^3}-\kappa.$$

We could solve this equation for $x$ to find the location of the minimum, but the algebra is a bit messy - this equation looks transcendental. We instead plot the potential and find the minimum by inspection. We can see the minimum is stable locally because the potential is concave up at the minimum. 
```

````python
def quarkPotential(gamma, delta, kappa, x):
    return -gamma/x+delta/x**2+kappa*x

xstart = 0.2
xend = 10
steps = 1000

x = np.linspace(xstart, xend, steps)

gamma = 10
delta = 3
kappa = 1

V = quarkPotential(gamma, delta, kappa, x)

minV = np.min(V)
minX = x[np.argmin(V)]

fig = plt.figure(figsize=(8, 6))

plt.plot(x, V, lw=2, label=r'$V(x)$')
plt.plot(minX, minV, 'C1o', label='Min. V={:.2e}V, x={:.2e}m'.format(minV, minX))
plt.plot(x, 0*x, 'k--')

plt.xlabel('x (m)')
plt.ylabel('V (J)')
plt.legend()
plt.grid(True)
````


````python
xstart = 0.2
xend = 200
steps = 10000

x = np.linspace(xstart, xend, steps)

gamma = 10
delta = 3
kappa = 1

Etot = 100

V = quarkPotential(gamma, delta, kappa, x)

fig = plt.figure(figsize=(8, 6))

plt.plot(x, V, lw=2, label=r'$V(x)$')
plt.plot(x, Etot*np.ones(steps), 'C1--', label=r'$E_{tot}=100\; \mathrm{J}$')

plt.xlabel('x (m)')
plt.ylabel('V (J)')
plt.legend()
plt.grid(True)
````

4. (20pt) We will now change our program (using either the standard Euler and the Euler-Cromer methods) and compute the position and the velocity as functions of time using the force computed from the above potential. Use as initial condition at time $t=0$ that the particle is $x_0=2$ and has initial velocity $v_0=0$. 

```{admonition} Solution
:class: tip

We start by writing an EulerCromer integrator for the force. We then use this to model the motion of the particle. We can plot the position and velocity of the particle as a function of time. We can also plot the total energy of the particle as a function of time. We can see that the energy is conserved in the simulation. 
```

````python
def IntegrateQuarkEOM_EC(gamma, delta, kappa, x0, v0, t0, tf, dt):
    t = np.arange(t0, tf, dt)
    x = np.zeros(t.shape)
    v = np.zeros(t.shape)
    x[0], v[0] = x0, v0

    for i in range(1, len(t)):
        v[i] = v[i-1] + dt * (-gamma/x[i-1]**2 + 2*delta/x[i-1]**3 - kappa)
        x[i] = x[i-1] + dt * v[i]

    return t, x, v

def IntegrateQuarkEOM_Euler(gamma, delta, kappa, x0, v0, t0, tf, dt):
    t = np.arange(t0, tf, dt)
    x = np.zeros(t.shape)
    v = np.zeros(t.shape)
    x[0], v[0] = x0, v0

    for i in range(1, len(t)):
        v[i] = v[i-1] + dt * (-gamma/x[i-1]**2 + 2*delta/x[i-1]**3 - kappa)
        x[i] = x[i-1] + dt * v[i-1]

    return t, x, v
````

````python

gamma = 10
delta = 3
kappa = 1

x0 = 2
v0 = 0

t0 = 0
tf = 10
N = 100000
dt = (tf - t0)/N

t, x, v = IntegrateQuarkEOM_EC(gamma, delta, kappa, x0, v0, t0, tf, dt)
Quark_EC = pd.DataFrame({'t': t, 'x': x, 'v': v})
t, x, v = IntegrateQuarkEOM_Euler(gamma, delta, kappa, x0, v0, t0, tf, dt)
QuarkEuler = pd.DataFrame({'t': t, 'x': x, 'v': v})

plt.figure(figsize=(10, 5))
plt.plot(Quark_EC.t, Quark_EC.x, label='x(t) (Euler-Cromer)')
plt.plot(QuarkEuler.t, QuarkEuler.x, label='x(t) (Euler)')
plt.xlabel('t (s)')
plt.ylabel('x (m)')
plt.legend()
plt.grid(True)

plt.figure(figsize=(10, 5))
plt.plot(Quark_EC.t, Quark_EC.v, label='v(t) (Euler-Cromer)')
plt.plot(QuarkEuler.t, QuarkEuler.v, label='v(t) (Euler)')
plt.xlabel('t (s)')
plt.ylabel('v (m/s)')
plt.legend()
plt.grid(True)
````

5. (10 pts) Test numerically that energy is conserved as function of time in either algorithm. 

```{admonition} Solution
:class: tip
In the plots below, we can see that the motion is oscillatory. We can also see that the energy is conserved in the Euler-Cromer simulation over one integration interval, but in the Euler integration it just grows.
```

````python
PE_EC = quarkPotential(gamma, delta, kappa, Quark_EC.x)
KE_EC = 0.5*Quark_EC.v**2
Etot_EC = PE_EC + KE_EC

PE_Euler = quarkPotential(gamma, delta, kappa, QuarkEuler.x)
KE_Euler = 0.5*QuarkEuler.v**2
Etot_Euler = PE_Euler + KE_Euler

plt.figure(figsize=(16, 12))

# Plot 1: PE comparison
plt.subplot(2, 2, 1)
plt.plot(Quark_EC.t, PE_EC, label='PE(t) (Euler-Cromer)')
plt.plot(QuarkEuler.t, PE_Euler, label='PE(t) (Euler)')
plt.xlabel('t (s)')
plt.ylabel('PE (J)')
plt.legend()
plt.grid(True)
plt.title('Potential Energy Comparison')

# Plot 2: KE comparison
plt.subplot(2, 2, 2)
plt.plot(Quark_EC.t, KE_EC, label='KE(t) (Euler-Cromer)')
plt.plot(QuarkEuler.t, KE_Euler, label='KE(t) (Euler)')
plt.xlabel('t (s)')
plt.ylabel('KE (J)')
plt.legend()
plt.grid(True)
plt.title('Kinetic Energy Comparison')

# Plot 3: Etot comparison
plt.subplot(2, 2, 3)
plt.plot(Quark_EC.t, Etot_EC, label='Etot(t) (Euler-Cromer)')
plt.plot(QuarkEuler.t, Etot_Euler, label='Etot(t) (Euler)')
plt.xlabel('t (s)')
plt.ylabel('Etot (J)')
plt.legend()
plt.grid(True)
plt.title('Total Energy Comparison')

# Plot 4: Euler-Cromer energies
plt.subplot(2, 2, 4)
plt.plot(Quark_EC.t, PE_EC, label='PE(t) (Euler-Cromer)')
plt.plot(Quark_EC.t, KE_EC, label='KE(t) (Euler-Cromer)')
plt.plot(Quark_EC.t, Etot_EC, label='Etot(t) (Euler-Cromer)')
plt.xlabel('t (s)')
plt.ylabel('Energy (J)')
plt.legend()
plt.grid(True)
plt.title('Euler-Cromer Energies')

plt.tight_layout()
plt.show()
````
