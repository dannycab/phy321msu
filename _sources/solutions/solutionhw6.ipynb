### Exercise 1 (10pt) Morse Potential as an SHO

If the potential has a local minimum, we can often find SHO approximation for that potential near the local minimum. 

The [Morse potential](https://en.wikipedia.org/wiki/Morse_potential) is a convenient model for the potential energy of a diatomic molecule. The potential is a radial one and thus one-dimensional. It is given by,

$$U(r) = A\left[ \left(e^{(R-r)/S}-1\right)^2-1\right]$$

where the distance between the centers of the two atoms is $r$, and the constants $A$, $R$, and $S$ are all positive. Here $S<<R$.

* 1a (2pt) Sketch (or plot) the potential as a function of $r$.

```{admonition} SOLUTION
:class: hint

We can plot the Morse potential, but we need to choose some values for the constants. Let's choose $A=1$, $R=1$, and $S=0.01$. 
```

````python
def MorsePotential(r, A=1, R=1, S=0.01):
    return A * (1 - np.exp(-S * (r - R)))**2

r = np.linspace(0.01, 3, 1000)
plt.plot(r, MorsePotential(r))
plt.xlabel('r')
plt.ylabel('V(r)')
plt.title('Morse Potential')
````


* 1b (3pt) Find the equilibrium position of the potential, i.e. the position where the potential is at a minimum. We will call this $r_e$.

```{admonition} SOLUTION
:class: hint

The equilibrium position is found by taking the derivative of the potential and setting it equal to zero. So we have,

$$\frac{dU}{dr} = 0 = -2A\left(e^{(R-r)/S}-1\right)\left(e^{(R-r)/S}\right)\left(-\frac{1}{S}\right).$$

We can solve for $r$ to find the equilibrium position.

$$0 = 2A\left(e^{(R-r)/S}-1\right)\left(e^{(R-r)/S}\right)\left(-\frac{1}{S}\right).$$

If we are looking for a finite solution, we focus on the first term in the product.

$$0 = \left(e^{(R-r)/S}-1\right),$$

$$1 = e^{(R-r)/S},$$

$$\ln(1) = \frac{R-r}{S},$$

$$0 = \frac{R-r}{S},$$

$$r_e = R.$$
```


* 1c (3pt) Rewrite the potential in terms of the displacement from equilibrium, $r = r_e + x$. Expand the potential to second order in $x$.

```{admonition} SOLUTION
:class: hint

We can expand the potential to second order in $x$ by using the Taylor expansion. First we plugin $r = r_e + x = R+x$ into the potential.

$$U(r_e+x) = A\left[ \left(e^{(R-(r_e+x))/S}-1\right)^2-1\right],$$

$$U(R+x) = A\left[ \left(e^{(R-R-x)/S}-1\right)^2-1\right],$$

$$U(x) = A\left[ \left(e^{-x/S}-1\right)^2-1\right].$$

So now we expand the potential to second order in $x$.

$$U(x) \approx A\left[ \left(1-\frac{x}{S}+\frac{x^2}{S^2}-1\right)^2-1\right],$$

$$U(x) \approx A\left[ \left(-\frac{x}{S} + \frac{x^2}{S^2}\right)^2-1\right],$$

$$U(x) \approx -A + A\left[ \frac{x^2}{S^2}\right].$$

```



* 1d (2pt) Find the effective spring constant, $k$, for the potential near the minimum. What is the frequency of small oscillations about the minimum?

```{admonition} SOLUTION
:class: hint

We can use the simple harmonic potential to find the effective spring constant. For the SHO,

$$U(x) = \frac{1}{2}kx^2 + U_0,$$

where $U_0$ is the potential at the minimum. We can compare this to the expansion of the approximate Morse potential to find the effective spring constant.

$$U(x) \approx -A + A\left[ \frac{x^2}{S^2}\right].$$

Such that,

$$U_0 = -A \quad \text{and} \quad k = \frac{2A}{S^2}.$$

For a particle of mass $m$, the frequency of small oscillations about the minimum is,

$$\omega = \sqrt{\frac{k}{m}} = \sqrt{\frac{2A}{mS^2}}.$$

```

### Exercise 2 (10pt), Simple Pendulum as an SHO

Consider a pendulum of length $L$ and mass $m$. It can swing back and forth in a plane, under the influence of gravity. The pendulum is designed so that it can swing through a full circle.

* 2a (3pt) Draw the free-body diagram for the pendulum, and write the equations of motion.

```{admonition} SOLUTION
:class: hint

The free-body diagram for the pendulum is shown below. 

![Free-body diagram for a pendulum](../images/assignments/sol_images/hw6_2a.png)


The forces acting on the pendulum are the tension in the string, $T$, and the force of gravity, $mg$. The angle it makes with the vertical is $\phi$. If we decompose the motion along the axis of the pendulum and perpendicular to it (the radial and tangential directions), we can write the equations of motion.

The radial equation is related the the net force, which is centripetal.

$$\sum F_r = T - mg\cos(\phi) = \dfrac{mv^2}{L},$$

where $v = L\frac{d\phi}{dt}$ is the velocity of the pendulum.  The tangential force governs the oscillations of the pendulum.

$$\sum F_t = -mg\sin(\phi) = mL\frac{d^2\phi}{dt^2}.$$


```


* 2b (3pt) Find the potential energy of the pendulum as a function of the angle $\phi$ it makes with the vertical. (Choose the zero of potential energy to be at the bottom of the pendulum's swing.)

```{admonition} SOLUTION
:class: hint

The potential energy of the pendulum is due to gravity. If we measure the angle from vertical, $\phi$, then at the bottom $\phi=0$, and $U=0$. At $\phi=\pi$, $U$ is a maximum. The function that does this is $\cos(\phi)$; it helps that it also captures the projected height properly as well. So the potential energy is,

$$U(\phi) = -mgL\cos(\phi).$$
```



* 2c (4pt) For small angles, show the potential reduces to an SHO. What is the effective spring constant, $k$, for the potential near the minimum? What is the frequency of small oscillations about the minimum?

```{admonition} SOLUTION
:class: hint

For small angles, $\cos(\phi) \approx 1 - \frac{\phi^2}{2}$, so the potential energy is,

$$U(\phi) \approx -mgL + \frac{1}{2}mgL\phi^2.$$

This is the potential energy of a simple harmonic oscillator. However, the units of the oscillator length are not quite right; we can multiply by $L/L$ and regroup the terms:

$$U(\phi) \approx -mgL + \frac{1}{2}m\frac{g}{L}\left(L\phi\right)^2 = U_0 + U_{spring},$$

$$U_{spring} = \frac{1}{2}m\frac{g}{L}\left(L\phi\right)^2 = \frac{1}{2}kx^2, \quad k = \frac{mg}{L}.$$

For a particle of mass $m$, the frequency of small oscillations about the minimum is,

$$\omega = \sqrt{\frac{k}{m}} = \sqrt{\frac{mg}{mL}} = \sqrt{\frac{g}{L}}.$$

Notice that the oscillatory variable is $L\phi$, which is the arc length of the pendulum. This is the effective length of the pendulum for small oscillations.
```

### Exercise 3 (10pt), Time Averaging and the SHO

Time Averaging is a common tool to use with periodic systems. It also us to discuss what happens to different properties of the system over one period.

An SHO has a period $\tau$. We can find the time average of a variable $f(t)$ over one period, $\langle f \rangle$, by averaging over the period,

$$\langle f \rangle = \frac{1}{\tau}\int_0^\tau f(t)dt$$

* 3a (2pt) Show that the time average of the position of the SHO is zero.

```{admonition} SOLUTION
:class: hint

Let's the take the general case of the SHO. It's position is given by,

$$x(t) = A\cos(\omega t - \delta),$$

where $A$ is the amplitude, $\omega$ is the angular frequency, and $\delta$ is the phase. The time average of the position can be constructed from the integral prescription above,

$$\langle x \rangle = \frac{1}{\tau}\int_0^\tau A\cos(\omega t - \delta)dt.$$

We convert the angular frequency to the period, $\omega = \frac{2\pi}{\tau}$, and the integral becomes,

$$\langle x \rangle = \frac{1}{\tau}\int_0^\tau A\cos\left(\frac{2\pi t}{\tau} - \delta\right)dt.$$

We now integrate the cosine function over one period and evaluate it,

$$\langle x \rangle = \frac{A}{\tau} \frac{\tau}{2\pi} \left[ \sin \left( \frac{2\pi t}{\tau} - \delta \right) \right]_0^\tau,$$

$$\langle x \rangle = \frac{A}{2\pi} \left[ \sin \left( 2\pi - \delta \right) - \sin \left( -\delta \right) \right].$$

So we are computing the difference of the sine function at two different points, but precisely one period apart ($2\pi$). The value of the function is the same at both points. So the time average of the position is zero,

$$\langle x \rangle = \frac{A}{2\pi} \left[ \sin \left( -\delta \right) - \sin \left( -\delta \right) \right],$$

$$\langle x \rangle = 0.$$

```


* 3b (2pt) Show that the time average of the velocity of the SHO is zero.

```{admonition} SOLUTION
:class: hint

Our solution from above is fairly general. The velocity of this SHO is,

$$v(t) = -A\omega\sin(\omega t - \delta).$$

We see that we will get an integral with the same form as before,

$$\langle v \rangle = \frac{1}{\tau}\int_0^\tau -A\omega\sin(\omega t - \delta)dt,$$

$$\langle v \rangle = \frac{-A\omega}{\tau}\int_0^\tau \sin\left(\frac{2\pi t}{\tau} - \delta\right)dt.$$

This integral evaluates no differently from the prior one. We take the difference of the integrated periodic function exactly one period apart. This will result in a zero time average as well. 

$$\langle v \rangle = 0.$$

```



* 3c (6pt) Show that the time average of the kinetic energy of the SHO is equal to the time average of the potential energy (and importantly, non-zero).  If the total energy is $E$, these time averages are equal to $E/2$. You might need to show that the very useful trigonometric identity,

$$\langle \sin^2(\omega t-\delta)\rangle = \langle \cos^2(\omega t-\delta)\rangle = \frac{1}{2}$$

```{admonition} SOLUTION
:class: hint

The energies of the SHO are,

$$K(t) = \frac{1}{2}m\left(-A\omega\sin(\omega t - \delta)\right)^2,$$

$$U(t) = \frac{1}{2}k\left(A\cos(\omega t - \delta)\right)^2.$$

So we know that we will have to compute the time average of the square of the sine and cosine functions. So let's just do that first,

$$\langle \sin^2(\omega t-\delta)\rangle = \frac{1}{\tau}\int_0^\tau \sin^2\left(\frac{2\pi t}{\tau} - \delta\right)dt.$$

We can use the trigonometric identity $\sin^2(x) = \frac{1}{2}(1 - \cos(2x))$ to simplify the integral,

$$\langle \sin^2(\omega t-\delta)\rangle = \frac{1}{\tau}\int_0^\tau \frac{1}{2}\left(1 - \cos\left(\frac{4\pi t}{\tau} - 2\delta\right)\right)dt.$$

We separate the integral into two parts,

$$\langle \sin^2(\omega t-\delta)\rangle = \frac{1}{2} - \frac{1}{2\tau}\int_0^\tau \cos\left(\frac{4\pi t}{\tau} - 2\delta\right)dt.$$

As before, we notice the integral of the cosine function over one period evaluates to zero. So are left with,

$$\langle \sin^2(\omega t-\delta)\rangle = \frac{1}{2}.$$

We can make a similar argument for the cosine function by using the trigonometric identity $\cos^2(x) = \frac{1}{2}(1 + \cos(2x))$. This would also give us,

$$\langle \cos^2(\omega t-\delta)\rangle = \frac{1}{2}.$$

And thus the time average of the kinetic and potential energies are,

$$\langle K \rangle = \frac{1}{2}m\omega^2A^2\langle \sin^2(\omega t-\delta)\rangle = \frac{1}{4}m\omega^2A^2,$$

$$\langle U \rangle = \frac{1}{2}kA^2\langle \cos^2(\omega t-\delta)\rangle = \frac{1}{4}kA^2.$$

We note that $k = m\omega^2$, so the time averages are equal,

$$\langle K \rangle + \langle U \rangle = \frac{1}{4}kA^2 + \frac{1}{4}kA^2 = \frac{1}{2}kA^2.$$

For an SHO with amplitude $A$ (as we constructed), the total energy is $E = \frac{1}{2}kA^2$, so the time averages are equal to $E/2$.

```

### Exercise 4 (10pt), Toy Potential

Consider a toy potential of the form,

$$U(r) = U_0\left(\dfrac{r}{R}+\lambda^2\frac{R}{r}\right)$$

where $U_0$, $R$, and $\lambda$ are all positive constants and the domain of the potential is $0<r<\infty$. 

* 4a (2pt) Sketch (or plot) the potential as a function of $r$.

```{admonition} SOLUTION
:class: hint

We can plot the toy potential, but we need to choose some values for the constants. Let's choose $U_0=1$, $R=1$, and $\lambda=0.1$. Thus,

$$U(r) = 1\left(\dfrac{r}{1}+0.1^2\frac{1}{r}\right).$$
```

````python
def ToyPotential(r, U0=1, R=1, lambda_=0.1):
    return U0*(r/R+lambda_**2*R/r)

r = np.linspace(0.01, 3, 1000)

fig = plt.figure(figsize=(10, 6))
plt.plot(r, ToyPotential(r))
plt.xlabel('r')
plt.ylabel('U(r)')
plt.title('Toy Potential')
plt.grid(True)
````



* 4b (3pt) Find the equilibrium position of the potential, i.e. the position where the potential is at a minimum. We will call this $r_e$.

```{admonition} SOLUTION
:class: hint

To find the equilibrium position, we take the derivative of the potential and set it equal to zero. So we have,

$$\frac{dU}{dr} = 0 = U_0\left(\dfrac{1}{R}-\lambda^2\frac{1}{r^2}\right).$$

We can solve for $r$ to find the equilibrium position.

$$0 = \dfrac{1}{R}-\lambda^2\frac{1}{r^2},$$

$$\lambda^2\frac{1}{r^2} = \dfrac{1}{R},$$

$$r^2 = \lambda^2R,$$

$$r_e = \pm\lambda R.$$

We take the positive root, $r_e = \lambda R$, as the equilibrium position because the potential is symmetric about the origin.
```

* 4c (5pt) Rewrite the potential in terms of the displacement from equilibrium, $r = r_e + x$. Expand the potential to second order in $x$. What is the effective spring constant, $k$, for the potential near the minimum? What is the frequency of small oscillations about the minimum?

```{admonition} SOLUTION
:class: hint

We can expand the potential to second order in $x$ by using the Taylor expansion. First we plugin $r = r_e + x = \lambda R + x$ into the potential.

$$U(r_e+x) = U_0\left(\dfrac{\lambda R + x}{R}+\lambda^2\frac{R}{\lambda R + x}\right).$$

We multiple this out to get,

$$U(r_e+x) = U_0\left(\lambda + \dfrac{x}{R}+\lambda^2\frac{R}{\lambda R + x}\right).$$

We can clean up the last term by taking out an $\lambda R$ from the denominator and then Taylor expanding the result,

$$U(r_e+x) = U_0\left(\lambda + \dfrac{x}{R}+\frac{\lambda^2R}{\lambda R}\frac{1}{1+\frac{x}{\lambda R}}\right),$$

$$U(r_e+x) = U_0\left(\lambda + \dfrac{x}{R}+\lambda \left(1-\frac{x}{\lambda R}+\frac{x^2}{(\lambda R)^2}-\frac{x^3}{(\lambda R)^3}+\cdots\right)\right).$$

Keeping only the terms up to second order in $x$,

$$U(x) \approx U_0\left(\lambda + \dfrac{x}{R}+\lambda -\frac{x}{R}+\frac{x^2}{(\lambda R)^2}\right),$$

$$U(x) \approx U_0\left(2\lambda + \frac{x^2}{(\lambda R)^2}\right).$$

Matching to the form of the SHO potential, we find the effective spring constant,

$$k = \frac{2U_0}{\lambda^2 R^2}.$$

For a particle of mass $m$, the frequency of small oscillations about the minimum is,

$$\omega = \sqrt{\frac{k}{m}} = \sqrt{\frac{2U_0}{m\lambda^2 R^2}}.$$
```

### Exercise 5 (10pt), Defining Periodicity

A common issue with oscillators is determining their periodicity. For the SHO, we can show that the period is $2\pi/\omega_0$ where $\omega_0$ is the natural frequency of the SHO. How might we define a periodicity more generally? Let's start with the damped harmonic oscillator. Consider a weakly damped oscillator ($\beta < \omega_0$). The motion of the oscillator is given by,

$$x(t) = A e^{-\beta t}\cos(\omega_1 t - \delta)$$

where $A$ is the amplitude, $\beta$ is the damping constant, $\omega_1 = \sqrt{\omega_0^2 - \beta^2}$  and $\delta$ is the phase.

The motion decays with time, but we can still define a periodicity, $\tau_1$, which is the time between peaks in the motion.

* 5a (4pt) Sketch (or plot) the motion of the oscillator as a function of time. Show that you can find $\tau_1 = 2\pi/\omega_1$ from looking at successive maxima.

```{admonition} SOLUTION
:class: hint

We can plot the motion of the oscillator, but we need to choose some values for the constants. Let's choose $A=1$, $\beta=0.1$, and $\omega_0=1$. Thus,

$$x(t) = e^{-0.1 t}\cos(\sqrt{1-0.01} t - \delta).$$

We can plot this function and look for the time between successive maxima. We can then compare this to an SHO with a frequency equal to $\omega_1$.

It's easy to see from the graph that each maximum in the damped oscillator occurs at the peak of the SHO with the same frequency. This is the definition of the period of the damped oscillator, $\tau_1 = 2\pi/\omega_1$.
```

````python
def DampedOscillator(t, A=1, omega=1, gamma=0.1, delta=0):
    return A * np.exp(-gamma * t) * np.cos(omega * t - delta)

def SHO(t, A=1, omega=1, delta=0):
    return A * np.cos(omega * t - delta)

t = np.linspace(0, 30, 1000)

fig = plt.figure(figsize=(10, 6))
plt.plot(t, DampedOscillator(t))
plt.plot(t, SHO(t))
plt.legend(['Damped', 'Simple'])
plt.grid(True)
plt.xlabel('t')
plt.ylabel('x(t)')
````


* 5b (3pt) Show that an equivalent definition of $\tau_1$ is twice the time between zero crossings of the motion.

```{admonition} SOLUTION
:class: hint

Again, from the graph, we can see that the time between zero crossings is half the time between successive maxima. This is simply another definition of the period of the damped oscillator, $\tau_1 = 2\pi/\omega_1$.
```


* 5c (3pt) If the damping $\beta$ is half the natural frequency ($\omega_0$), how does the amplitude of the motion decay in one period?

```{admonition} SOLUTION
:class: hint

Using the general solution,

$$x(t) = A e^{-\beta t}\cos(\omega_1 t - \delta),$$

we can see that using the definition of $\omega_1 = \sqrt{\omega_0^2 - \beta^2}$, we can write the motion as,

$$x(t) = A e^{-\beta t}\cos(\sqrt{\omega_0^2 - \beta^2} t - \delta).$$

With $\beta = \omega_0/2$, we have,

$$x(t) = A e^{-\omega_0 t/2}\cos\left(\sqrt{\omega_0^2 - \left(\frac{\omega_0}{2}\right)^2} t - \delta\right),$$

$$x(t) = A e^{-\omega_0 t/2}\cos\left(\sqrt{\frac{3\omega_0^2}{4}} t - \delta\right).$$

From this form, we have found the period of this damped oscillator,

$$\tau_1 = \frac{2\pi}{\sqrt{3\omega_0^2/4}} = \frac{4\pi}{\sqrt{3}\omega_0}.$$

And thus the amplitude decays thusly,

$$A e^{-\omega_0 \tau_1/2} = A e^{-2\pi/\sqrt{3}}.$$
```

#### Examples of 1D potentials

1. **Simple Pendulum (Nonlinear Small Angle Approximation)**:
   - Potential: $ V(\theta) = mgh(1 - \cos(\theta)) $, where $m$ is the mass, $g$ is the acceleration due to gravity, $h$ is the length of the pendulum, and $\theta$ is the angular displacement.
2. **Nonlinear Spring (Hardening or Softening)**:
   - Potential: $ V(x) = \frac{k}{2} x^2 + \frac{\beta}{3} x^3 $, where $k$ and $\beta$ are constants. Depending on the sign of $\beta$, the spring can exhibit hardening ($\beta > 0$) or softening ($\beta < 0$) nonlinearity.
3. **Lennard-Jones Potential Oscillator** (for a diatomic molecule model):
   - Potential: $ V(r) = 4\epsilon \left[ \left(\frac{\sigma}{r}\right)^{12} - \left(\frac{\sigma}{r}\right)^6 \right] $, where $\epsilon$ is the depth of the potential well, $\sigma$ is the finite distance at which the inter-particle potential is zero, and $r$ is the distance between particles.
4. **Morse Potential (for molecular vibrations)**:
   - Potential: $ V(x) = D_e \left(1 - e^{-a(x - x_0)}\right)^2 $, where $D_e$ is the depth of the potential well, $a$ is a constant related to the width of the well, $x$ is the displacement from equilibrium, and $x_0$ is the equilibrium bond length. This potential models the energy of a diatomic molecule as a function of the distance between atoms, showing oscillatory behavior that represents molecular vibrations.
5. **Double Well Potential**:
   - Potential: $ V(x) = -\frac{\mu}{2} x^2 + \frac{\lambda}{4} x^4 $, where $\mu$ and $\lambda$ are positive constants. This system exhibits bistability with two stable equilibria, leading to interesting nonlinear dynamics and potential oscillations between the wells under certain conditions.
